```python
import pandas as pd
import numpy as np
from pandas.core import sample
```

```python
df = pd.read_csv("house_price_increase.csv")
df.head()
```

```python
df.shape
```

```python
pop_mean = 10
pop_std_dev = 4
```

```python
sample_mean = df.price_increase_pct.mean()
sample_mean
```

```python
std_err = pop_std_dev / np.sqrt(df.shape[0])
std_err
```

```python
z_score = (sample_mean - pop_mean) / std_err
z_score
```

```python
alpha = 0.05 # alpha means significance level
```

```python
from scipy import stats
z_critical = stats.norm.ppf(1 - alpha)
z_critical
```

```python
z_score, z_critical
```

### z_score > z_critical it means that we will reject the NULL hypothesis. This means the alternate hypothesis becomes true
