```python
class User:
    def __init__(self, user_id, name, age):
        self.user_id = user_id
        self.name = name
        self.age = age
```

```python
dhaval = User(121, "Dhaval", 34)
dhaval.age
```

```python
from pydantic import BaseModel

class User1(BaseModel):
    id : int
    name : str
    age : int

dhanu = User1(id = 121, name = "Dhaval", age = 34)
dhanu.name
dhanu.age
dhanu.id

```
