```python
import time
```

```python
def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Function {func.__name__} took {(end - start) * 1000} milliseconds")
        return result
    return wrapper
```

```python
@timer
def cube_list(number):
    result = []
    for i in number:
        result.append(i ** 3)
    return result

_ = cube_list(range(1000000))
```

```python
@timer
def square_list(number):
    result = []
    for i in number:
        result.append(i ** 2)
    return result
_ = square_list(range(1000000))
```
