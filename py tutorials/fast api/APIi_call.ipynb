```python
import requests as r

url = "https://jsonplaceholder.typicode.com/posts/2"

response = r.get(url)

response
```

```python
response.status_code
```

```python
if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print(response.status_code)
    print("Some error occurred")
```
