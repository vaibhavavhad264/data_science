```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("miles.csv")
df.head()
```

```python
mean_pop = df.miles.mean()
std_pop = df.miles.std()
mean_pop, std_pop
```

```python
plt.figure(figsize= (8, 4))
plt.plot(df.miles)
```

```python
sns.histplot(df.miles, kde= True)
```

```python
SAMPLE_SIZE = 50
sample = df.sample(SAMPLE_SIZE, random_state = 100)
sample.head()
```

```python
mean_sample = sample.miles.mean()
std_sample = sample.miles.std()

mean_sample, std_sample
```

```python
import scipy.stats as st

z_score_95_confidence = st.norm.ppf(0.975)
z_score_95_confidence
```

```python
import math

margin_error = z_score_95_confidence * (std_pop/np.sqrt(SAMPLE_SIZE))
margin_error
```

```python
lower = mean_sample - margin_error
upper = mean_sample + margin_error

(lower, upper)
```
