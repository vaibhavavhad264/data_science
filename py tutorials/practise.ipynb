```python

class PlayerInfo:
    def __init__(self, fname, lname, age, team):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.team = team
        self.score = []
    def add_score(self, score):
        self.score.append(score)
    def average_score(self):
        self.average = sum(self.score)/len(self.score)
        return self.average
    def __str__(self):
        return f"{self.fname} {self.lname} ia an {self.team} cricketer. And his average score is {self.average}."
p1 = PlayerInfo("Virat", "Kohli", 39, "INDIA")
p1.add_score(67)
p1.add_score(59)
p1.add_score(77)
p1.add_score(103)
print(p1.average_score())

```
