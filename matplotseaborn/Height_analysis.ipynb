```python
import pandas as pd
```

```python
df = pd.read_csv("Height main data.csv")
df.head(2)
```

```python
Q1, Q3 = df.Height.quantile([0.25, 0.75])
round(Q1,2), round(Q3,2)
```

```python
IQR = round(Q3-Q1, 3)
IQR
```

```python
lower_limit = round(Q1 - IQR * 1.5, 3)
upper_limit = round(Q3 + IQR * 1.5, 3)
lower_limit, upper_limit
```

```python
df_new = df[(df.Height > lower_limit) & (df.Height < upper_limit)]
df_new
```
