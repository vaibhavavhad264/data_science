```python
import pandas as pd
from matplotlib import pyplot as plt
from plotly.validators.layout.geo.projection import rotation
from typer import colors
```

```python
adidas_df = df[df["Brand"] == "Adidas"]
adidas_df.head()
```

```python
adidas_df.shape
```

```python
adidas_df.Units_Sold.sum()
```

```python
def plot_qty():
    plt.figure(figsize = (15, 6))

    dates = nike_df["Date"]
    plt.plot(dates, nike_df["Units_Sold"], marker = "o", label = "Nike", color = "Blue")
    plt.plot(dates, adidas_df["Units_Sold"], marker = "o", label = "Adidas", color = "Red")
    plt.xlabel("Date")
    plt.ylabel("Total Sold Quantity")
    plt.title("Daily Sales Qty for Nike and Adidas")
    plt.xticks(rotation = 45)
    plt.legend()
    plt.tight_layout()
    plt.grid(True, which = "both", linestyle = "--", linewidth = 0.5)
    plt.show()
```

```python
plot_qty()
```
