```python
import pandas as pd
from matplotlib.pyplot import xlabel

df = pd.read_excel("Book1.xlsx")
df
```

```python
from matplotlib import pyplot as plt

plt.figure(figsize = (12, 4))
plt.plot(df['Quarter'], df['Fridge'], color = 'blue', label = 'Fridge')
plt.plot(df['Quarter'], df['Dishwasher'], color = 'green', label = 'Dishwasher')
plt.plot(df['Quarter'], df['Washing Machine'], color = 'orange', label = 'Washing Machine')

plt.title('Product Sales')
plt.ylabel('Revenue (mln $)')
plt.xlabel('Quarter')

plt.legend()
plt.show()
```

```python
total_sale = df[['Fridge', 'Dishwasher', 'Washing Machine']].sum()
total_sale
```

```python
total_sale.index
```

```python
plt.pie(total_sale, labels=['Fridge', 'Dishwasher', 'Washing Machine'], autopct = '%1.2f%%',
        explode = (0.05, 0.05, 0.05), shadow = True, startangle = 300)
plt.show()
```
