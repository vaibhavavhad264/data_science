```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heights.csv")
df.head()
```

```python
df.describe()
```

```python
mean = round(df.height.mean(), 2)
std = round(df.height.std(), 2)

mean, std
```

```python
sns.histplot(df.height, kde = True)
plt.axvline(mean, color = 'black', linewidth = 1, linestyle = '--')

plt.axvline(mean - std , color = 'red', linewidth = 1, linestyle = '--')
plt.axvline(mean + std, color = 'red', linewidth = 1, linestyle = '--')

plt.axvline(mean - std*2 , color = 'g', linewidth = 1, linestyle = '--')
plt.axvline(mean + std*2, color = 'g', linewidth = 1, linestyle = '--')

plt.axvline(mean - std*3 , color = 'b', linewidth = 1, linestyle = '--')
plt.axvline(mean + std*3, color = 'b', linewidth = 1, linestyle = '--')

plt.show()
```

```python
l1 = round(mean - std , 2)
r1 = round(mean + std , 2)
l1, r1
```

```python
l2 = round(mean - std*2, 2)
r2 = round(mean + std*2, 2)
l2, r2
```

```python
l3 = round(mean - std*3 , 2)
r3 = round(mean + std*3 , 2)
l3, r3
```
