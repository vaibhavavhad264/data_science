```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heights_with_outliers.csv")
df.head()
df.shape
```

```python
mean =round( df.height.mean(), 2)
std = round(df.height.std(), 2)
mean, std
```

```python
def hist_analysis():
    sns.histplot(df.height, kde = True)
    plt.axvline(mean, color = 'black', linewidth = 1, linestyle = '--')

    plt.axvline(mean - std , color = 'red', linewidth = 1, linestyle = '--')
    plt.axvline(mean + std, color = 'red', linewidth = 1, linestyle = '--')

    plt.axvline(mean - std*2 , color = 'g', linewidth = 1, linestyle = '--')
    plt.axvline(mean + std*2, color = 'g', linewidth = 1, linestyle = '--')

    plt.axvline(mean - std*3 , color = 'b', linewidth = 1, linestyle = '--')
    plt.axvline(mean + std*3, color = 'b', linewidth = 1, linestyle = '--')

    plt.show()
    return 0
```

```python
hist_analysis()
```

```python
lower = mean - 3*std
upper = mean + 3*std
lower, upper
```

```python
df[(df.height < lower) | (df.height > upper)]
```

```python
df_new = df[df.height.between(lower, upper)]
df_new.shape
```

```python
sns.histplot(df_new, kde = True)
```
